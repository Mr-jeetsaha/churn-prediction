```python
import tensorflow as tf
from tensorflow.keras.models import load_model  #type:ignore
import pickle
import pandas as pd
```

```python
# load the trained model, scaler, one-hot-encoding, label-encoder
model=load_model('model.h5')

with open('onehot_encoder_geo.pkl', 'rb') as file:
    label_encoder_geo=pickle.load(file)
    
with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender=pickle.load(file)
    
with open('scaler.pkl', 'rb')as file:
    scaler=pickle.load(file)
```

```python
# Example input data
input_data={
    'CreditScore':50000,
    'Geography':'France',
    'Gender':'Female',
    'Age':30,
    'Tenure':4,
    'Balance':60000,
    'NumOfProducts':3,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}

input_data
```

```python
# One-hot encoding for 'Geography'
geo_encoded=label_encoder_geo.transform([[input_data['Geography']]]).toarray()
geo_encoded_df=pd.DataFrame(geo_encoded, columns=label_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
# Here input data is in the key_value format, so we have to convert it into dataframe 
input_df=pd.DataFrame([input_data])
input_df
```

```python
# Encode categorical variable (Gender column)
input_df['Gender']=label_encoder_gender.transform(input_df['Gender'])
input_df
```

```python
# Concatination of one hot encoded and drop the existing geography column
input_df=pd.concat([input_df.drop("Geography", axis=1), geo_encoded_df],axis=1)
input_df
```

```python
# Scaling the input data
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
prediction=model.predict(input_scaled)
prediction
```

```python
# probability of the prediction
prediction_prob=prediction[0][0]
prediction_prob
```

```python
if prediction_prob>0.5:
    print("The customer is likely to churn")
else:
    print("The customer is not likely to churn")
```
